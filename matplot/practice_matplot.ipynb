```python
# 1 - simple line graph
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Simple Line Graph")
plt.show()
```

```python
# 2 -  Program to draw a line using given axis values
x = [1,2,3]
y = [2,4,1]
plt.plot(x,y)
plt.xlabel("X axis values")
plt.ylabel("Y axis values")
plt.title("Line graph using axis values")
plt.show()
```

```python
# 3 - Draw a line using given axis values taken from a text file
x = []
y = []
with open('test_data.txt','r') as file:
    for line in file:
        values = line.split()
        x.append(int(values[0]))
        y.append(int(values[1]))
        
plt.plot(x,y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Graph from text file")
plt.show()
```

```python
#  line charts of the financial data of Alphabet Inc
data = {
    'Date': ['03-10-16', '04-10-16', '05-10-16', '06-10-16', '07-10-16'],
    'Open': [774.25, 776.030029, 779.309998, 779, 779.659973],
    'High': [776.065002, 778.710022, 782.070007, 780.47998, 779.659973],
    'Low': [769.5, 772.890015, 775.650024, 775.539978, 770.75],
    'Close': [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]
}
df = pd.DataFrame(data)
plt.plot(df['Date'],df['Open'],label = 'Open')
plt.plot(df['Date'], df['High'], label='High')
plt.plot(df['Date'], df['Low'], label='Low')
plt.plot(df['Date'], df['Close'], label='Close')
plt.xlabel('Date')
plt.ylabel('Price')
plt.title('Alphabet Inc. Financial Data (Oct 3–7, 2016)')
plt.legend()
plt.show()
```

```python
# 5 -  program to plot two or more lines on same plot with suitable legends of each line
x = [1,2,3,4,5]
y1 = [2,4,6,8,10]
y2 = [1,3,5,7,9]
y3 = [1,4,9,16,25]
plt.plot(x,y1, label = 'Even Numbers')
plt.plot(x,y2, label = "Odd Numbers")
plt.plot(x,y3, label = 'Squares')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Mutiple lines on Same Plot")
plt.legend()
plt.show()
```

```python
# 6 - Program to plot two or more lines with legends, different widths and colors.
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [8,16,24,32,40]
y3 = [5,10,15,20,25]
plt.plot(x,y1, color = 'blue', linewidth = 8, label='Line-1')
plt.plot(x,y2, color = 'red', linewidth = 1,label='Line-2')
plt.plot(x,y3, color = 'green', linewidth = 4,label='Line-3')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Lines with Different Colors and Widths')
plt.legend()
plt.show()
```

```python
# 7 -  Python program to plot two or more lines with different styles 
x = [1, 2, 3, 4, 5]
y1 = [5, 10, 15, 20, 25]
y2 = [3, 6, 9, 12, 15]
y3 = [25, 20, 15, 10, 5]
plt.plot(x, y1, linestyle='-', label='Solid Line')
plt.plot(x, y2, linestyle='--', label='Dashed Line')
plt.plot(x, y3, linestyle=':', label='Dotted Line')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Lines with Different Line Styles')
plt.legend()
plt.show()
```

```python
# 8 -  Python program to plot two or more lines and set the line markers.
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [5,10,15,20,25]
y3 = [3,6,9,12,15]
plt.plot(x,y1,marker='o',label='Circle marker')
plt.plot(x,y2,marker='s',label='Square marker')
plt.plot(x,y3,marker='^',label='Triangle marker')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Lines with Different Markers')
plt.legend()
plt.show()
```

```python
# 9 -  Python program to display the current axis limits values and set new axis values.
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y)
x_lim = plt.xlim()
y_lim = plt.ylim()
print("Current X-axis limits:", x_lim)
print("Current Y-axis limits:", y_lim)
plt.xlim(0,6)
plt.ylim(0,60)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Displaying and Setting Axis Limits')
plt.show()
```

```python
# 13 - Customized the grid lines with linestyle -, width .5. and color blue.
dates = ['03-10-16', '04-10-16', '05-10-16', '06-10-16', '07-10-16']
values = [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]
plt.plot(dates,values)
plt.xlabel("Date")
plt.ylabel("Closing Value")
plt.title("Alphabet Inc. Closing Prices (Oct 3–7, 2016)")
plt.grid(linestyle = '-',linewidth=0.5,color='red')
plt.show()
```

```python
# 14 - Line Chart with Major & Minor Grid
dates = ['03-10-16', '04-10-16', '05-10-16', '06-10-16', '07-10-16']
values = [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]
plt.plot(dates,values,marker = 'o')
plt.xlabel("Date")
plt.ylabel("Closing Value")
plt.title("Alphabet Inc. Closing Prices (Oct 3–7, 2016)")
# plt.minorticks_on() → enables minor grid lines
plt.minorticks_on()
plt.grid(which='major', linestyle='-', linewidth=0.8)
plt.grid(which='minor', linestyle=':', linewidth=0.2)
plt.tick_params(axis='both', which='both', length=0)# not oof - Tick marks will be visible
plt.show()
```

```python
# Bar Chart
# 1 - To display a bar chart of the popularity of programming Languages.
languages = ['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8.0, 7.7, 6.7]
plt.bar(languages, popularity)
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
plt.show()
```

```python
# 2 - Create horizontal bar chart
plt.barh(languages, popularity)
plt.xlabel('Popularity (%)')
plt.ylabel('Programming Languages')
plt.title('Popularity of Programming Languages')
plt.show()
```

```python
# 3 - Create bar chart with uniform color
plt.bar(languages, popularity, color='green')
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
plt.show()
```

```python
# 4 - Different colors for each bar
colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown']
plt.bar(languages, popularity, color=colors)
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
plt.show()
```

```python
# Add text label above each bar
bars = plt.bar(languages, popularity)
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
# bars = plt.bar(...) → stores each bar object
# bar.get_height() → gets value of each bar
# plt.text(x, y, value) → prints value above bar
# ha='center' → center align text
# va='bottom' → place text just above the bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height,
                f'{height}', ha='center', va='bottom')
plt.show()
```

```python
# 6 -  Create bar chart with blue border
plt.bar(languages, popularity, edgecolor='blue')
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
plt.show()
```

```python
# 7 - Specify positions for bars
positions = [0, 1, 2, 3, 4, 5]
plt.bar(positions, popularity)
# Set language names at given positions
plt.xticks(positions, languages)
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
plt.show()
```

```python
# 8 - # Positions and width
positions = [0, 1, 2, 3, 4, 5]
bar_width = 0.5
plt.bar(positions, popularity, width=bar_width)
# Set labels at positions
plt.xticks(positions, languages)
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
plt.show()
```

```python
# 9 - Increase bottom margin
plt.bar(languages, popularity)
plt.subplots_adjust(bottom=0.25)
plt.xlabel('Programming Languages')
plt.ylabel('Popularity (%)')
plt.title('Popularity of Programming Languages')
plt.show()
```
