```python
# Scatter plot for random 1000 x and y coordinates
import random
import plotly.graph_objects as go
x = [random.randint(1,100) for i in range(100)]
y = [random.randint(1,100) for i in range(100)]
# go.Figure() creates an empty graph container.
fig = go.Figure()
# add_trace() is used to add data (graph) to the figure.
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='markers',
    name='Random Points'
))
fig.update_layout(title='Scatter plot with 100 random points')
fig.show()
```

```python
#  line and scatter plots for random 100 x and y coordinates
x = [random.randint(1,100)for i in range(100)]
y = [random.randint(1,100)for i in range(100)]
fig = go.Figure()
# Line Plot
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='lines',
    name='Line Plot'
))
# Scatter Plot
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='markers',
    name='Scatter Plot'
))
fig.update_layout(
    title="Line and Scatter Plot with 100 Random Points",
    xaxis_title="X values",
    yaxis_title="Y values"
)
fig.show()
```
