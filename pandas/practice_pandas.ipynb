```python
# 1 - One dimensional array using pandas
import pandas as pd
import numpy as np
data = [10,20,30,40,50]
arr = pd.Series(data)
print(arr)
```

```python
# 2 - Series into python list
s = pd.Series([10,20,30,40])
py_list = s.tolist()
print(py_list)
print(type(py_list))
```

```python
# 3 - Addition,subtraction and two pandas series
s1 = pd.Series([2,4,6,8,10])
s2 = pd.Series([1,3,5,7,9])
addition = s1 + s2
print("Addition")
print(addition)
subtraction = s1 - s2
print("Subtraction")
print(subtraction)
multiplication = s1 * s2
print("Multiplication")
print(multiplication)
print("Division")
division = s1/s2
print(division)
```

```python
# 4 - Python program to get the powers of an array values
s1 = pd.Series([0, 1, 2, 3, 4, 5, 6])
s2 = pd.Series([1, 1, 3, 3, 3, 3, 3])
result = s1 ** s2
print("Original array")
print(s1.values)
print("First array elements raised to powers from second array, element-wise:")
print(result.values)
```

```python
# 10 - Select the rows where the number of attempts in the examination is greater than 2.
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily',
                'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no',
                'yes', 'yes', 'no', 'no', 'yes']
}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
df = pd.DataFrame(exam_data,index=labels)
# one method
res = df[df['attempts']> 2]
print(res)
# another method
res1 =df.loc[df['attempts'] > 2]
res1
```

```python
# 11 - Count the no of rows and columns of a data frame
rows = df.shape[0]
# shape[0] - gives rows count
# shape[1] - gives column count
columns = df.shape[1]
print(f"Rows count :{rows},Columns count :{columns}")
```

```python
# 12 -  Program to select the rows where the score is missing, i.e. is NaN
# isnull() - to find out the null values
missing = df[df['score'].isnull()]
print(missing)
# another method
mis = df.loc[df['score'].isna()]
print(mis)
# another method
miss = df[df['score'].isna()]
print(miss)
```

```python
# 13 -  Program to select the rows where number of attempts in the examination is less than 2 and score greater than 15.
result = df[(df['attempts'] < 2) & (df['score'] > 15)]
result
```

```python
# 14 - Python program to change the score in row 'd' to 11.5.
df.loc['d','score'] = 11.5
res = df.loc['d']
res
```

```python
# 15 - Program to calculate the sum of the examination attempts by the students.
total = df['attempts'].sum()
# sum() - Automatically ignores missing values (NaN)
print(total)
```
