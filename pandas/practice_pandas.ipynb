```python
# 3 - Addition,subtraction and two pandas series
s1 = pd.Series([2,4,6,8,10])
s2 = pd.Series([1,3,5,7,9])
addition = s1 + s2
print("Addition")
print(addition)
subtraction = s1 - s2
print("Subtraction")
print(subtraction)
multiplication = s1 * s2
print("Multiplication")
print(multiplication)
print("Division")
division = s1/s2
print(division)
```

```python
# 4 - Python program to get the powers of an array values
s1 = pd.Series([0, 1, 2, 3, 4, 5, 6])
s2 = pd.Series([1, 1, 3, 3, 3, 3, 3])
result = s1 ** s2
print("Original array")
print(s1.values)
print("First array elements raised to powers from second array, element-wise:")
print(result.values)
```

```python
# 10 - Select the rows where the number of attempts in the examination is greater than 2.
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily',
                'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no',
                'yes', 'yes', 'no', 'no', 'yes']
}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
df = pd.DataFrame(exam_data,index=labels)
# one method
res = df[df['attempts']> 2]
print(res)
# another method
res1 =df.loc[df['attempts'] > 2]
res1
```

```python
# 11 - Count the no of rows and columns of a data frame
rows = df.shape[0]
# shape[0] - gives rows count
# shape[1] - gives column count
columns = df.shape[1]
print(f"Rows count :{rows},Columns count :{columns}")
```

```python
# 12 -  Program to select the rows where the score is missing, i.e. is NaN
# isnull() - to find out the null values
missing = df[df['score'].isnull()]
print(missing)
# another method
mis = df.loc[df['score'].isna()]
print(mis)
# another method
miss = df[df['score'].isna()]
print(miss)
```

```python
# 13 -  Program to select the rows where number of attempts in the examination is less than 2 and score greater than 15.
result = df[(df['attempts'] < 2) & (df['score'] > 15)]
result
```

```python
# 14 - Python program to change the score in row 'd' to 11.5.
df.loc['d','score'] = 11.5
res = df.loc['d']
res
```

```python
# 15 - Program to calculate the sum of the examination attempts by the students.
total = df['attempts'].sum()
# sum() - Automatically ignores missing values (NaN)
print(total)
```

```python
# 16 - Program to calculate the mean score for each different student in DataFrame.
# groupby()- is used to split data into groups, apply a function, and combine the results.
# df.groupby('column_name')['target_column'].aggregation_function()
mean_per_student = df.groupby('name')['score'].mean()
mean_per_student
```

```python
# 19 -  Write a Python program to replace the 'qualify' column contains the values 'yes' and 'no' with True and False.
# recreating the data frame again
# it shows warning reason (converting object to bool)
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily',
                'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no',
                'yes', 'yes', 'no', 'no', 'yes']
}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
df = pd.DataFrame(exam_data, index=labels)
print("'Before update:'")
res = df['qualify']
print(res)
df['qualify'] = df['qualify'].replace({'yes':True,'no':False})
df
```

```python
# 20 - Program to delete the 'attempts' column from the DataFrame.
# axis = 1   -it will delete column
# axis = 0   -it will delete row
print("'Before delete :'")
print(df)
df = df.drop('attempts',axis=1)
print("'After delete:'")
print(df)
```

```python
# 21 - Insert a column into DataFrame
df['attempts']=[1, 3, 2, 3, 2, 3, 1, 1, 2, 1]
print("'After insert a column'")
print(df)
```

```python
# 22 - Program to iterate over rows in a DataFrame.
# iterrows() - most common method to iterate over rows
for index,row in df.iterrows():
    print(index,row['name'],row['score'],row['qualify'],row['attempts'])
# Another method
# using itertuples() - faster
print("'Another method :'\n")
for row in df.itertuples():
    print(row.Index,row.name,row.score,row.qualify,row.attempts)
```
